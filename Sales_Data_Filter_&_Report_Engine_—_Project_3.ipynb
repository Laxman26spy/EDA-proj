```python
import numpy as mn

months = mn.arange(1,13)

mn.random.seed(42)
sales = mn.random.randint(10000, 20001, size=(10, 12))

print("Months:", months)
print("Sales Data Shape:", sales.shape)
print("\nSales Data:\n", sales)
```

mn.arange(1, 13) creates month numbers from 1 to 12.

mn.random.randint(10000, 20001, size=(10, 12)) generates sales figures for 10 stores across 12 months.


sales.shape should return (10, 12).

```python
S1 = sales[:,0:3]
S2 = sales[:,3:6]
S3 = sales[:,6:9]
S4 = sales[:,9:12]

print("S1 Shape:",S1.shape)
print("S2 Shape:",S2.shape)
print("S3 Shape:",S3.shape)
print("S4 Shape:",S4.shape)

store_1 = sales[0, :]
print("\nStore 1 Monthly Sales:", store_1)

print("\nStore 1 Q1 Sales:", sales[0, 0:3])

print("\nQuarterly Total Sales:")


print("S1:", mn.sum(S1))
print("S2:", mn.sum(S2))
print("S3:", mn.sum(S3))
print("S4:", mn.sum(S4))
```

sales[:, 0:3] selects all stores and the first three months.


sales[0, :] selects all months for the first store.


sales[0, 0:3] selects the first store's sales for January–March.




```python
target = 15000

high_sales_mark = sales > target

print("Number of Store-Months Above Target:",
mn.sum(high_sales_mark))

store_averages = mn.mean(sales, axis=1)
overall_average = mn.mean(sales)

below_average_stores = store_averages < overall_average

print("\nOverall Average Sales:", overall_average)

print("Below-Average Store Indices:",
mn.where(below_average_stores)[0])

combined_and = high_sales_mark & below_average_stores[:, mn.newaxis]

print("\nAbove-Target Sales in Below-Average Stores:")
print(sales[combined_and])

combined_or = high_sales_mark | below_average_stores[:, mn.newaxis]

print("\nNumber of Results Using OR:",
mn.sum(combined_or))





```

sales > target produces a Boolean array containing True and False.


np.mean(sales, axis=1) calculates the average monthly sales for each store.


& means both conditions must be true.


| means at least one condition must be true.


np.where() returns the indices where a condition is true.

```python
selected_indices = mn.array([1, 4, 7])
selected_sales = sales[selected_indices, :]

print("Selected Stores: 2, 5, 8")
print("\nSelected Stores' Sales:\n", selected_sales)

top_three_indices = mn.argsort(selected_sales, axis=1)[:, -3:]

print("\nBest 3 Months for Each Selected Store:")

for i, store_index in enumerate(selected_indices):
    best_indices = top_three_indices[i]
    best_sales = sales[store_index, best_indices]

    order = mn.argsort(best_sales)[::-1]
    best_indices = best_indices[order]
    best_sales = best_sales[order]

    print(f"\nStore {store_index + 1}:")

    for month_index, amount in zip(best_indices, best_sales):
        print(f"Month {month_index + 1}: Sales = {amount}")



```

sales[selected_indices, :] selects multiple rows at once.


np.argsort() finds indices that would sort values.


[:, -3:] extracts the indices of the three largest monthly sales values for each selected store.


store_index + 1 displays human-friendly store numbers.

```python
monthly_totals = mn.sum(sales, axis=0)

print("Monthly Totals:")
for month, total in zip(months, monthly_totals):
 print(f"Month {month}: Total Sales = {total}")

store_totals = mn.sum(sales, axis=1)
store_averages = mn.mean(sales, axis=1)

growth_rates = mn.linspace(0.05, 0.15, 10)

growth_targets = store_averages * (1 + growth_rates)

print("\nSTORE-WISE-REPORTS")

for i in range(10):
  print(f"\nStore {i + 1}")
  print(f"Annual Sales: {store_totals[i]:.0f}")
  print(f"Average Monthly Sales: {store_averages[i]:.2f}")
  print(f"Growth Rate: {growth_rates[i] * 100:.2f}%")
  print(f"Monthly Growth Target: {growth_targets[i]:.2f}")

if store_averages[i] >= growth_targets[i]:
        print("Status: Target Achieved")
else:
        print("Status: Target Not Achieved")

best_month_index = mn.argmax(monthly_totals)

print("\n FINAL-SUMMARY")
print("Best Sales Month:", best_month_index + 1)
print("Highest Monthly Sales:", monthly_totals[best_month_index])
print("Total Annual Sales:", mn.sum(sales))
```

np.linspace(0.05, 0.15, 10) generates 10 evenly spaced growth rates, starting at 5% and ending at 15%.

The growth target is a monthly average sales target, not an annual sales target.

 Since we compare each store's average monthly sales against that target, the report uses consistent units.
